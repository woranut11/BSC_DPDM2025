### ขั้นตอนที่ 1: ระบุช่วงวันที่สำหรับแต่ละสถานี

### ขั้นตอนที่ 2: คำนวณจำนวนวันที่คาดว่าจะมีข้อมูลสำหรับแต่ละสถานี

**เหตุผล:** การคำนวณจำนวนวันที่คาดว่าจะมีข้อมูลในแต่ละสถานี (จากวันที่เริ่มต้นถึงวันที่สิ้นสุด) เป็นสิ่งสำคัญในการหาจำนวนข้อมูลที่ขาดหายไป โดยการนำจำนวนวันที่คาดหวังนี้ไปเปรียบเทียบกับจำนวนข้อมูลจริงที่มีอยู่

````python
date_ranges['expected_days'] = (date_ranges['max'] - date_ranges['min']).dt.days + 1
display(date_ranges.head())
````

### ขั้นตอนที่ 3: นับจำนวนบันทึกข้อมูลจริงสำหรับแต่ละสถานี

**เหตุผล:** เพื่อคำนวณจำนวนบันทึกข้อมูลที่หายไป จำเป็นต้องทราบจำนวนบันทึกข้อมูลจริงที่มีอยู่สำหรับแต่ละสถานี การนับจำนวนบันทึกข้อมูลจริงนี้จะนำไปเปรียบเทียบกับจำนวนวันที่คาดหวังเพื่อหาความสมบูรณ์ของข้อมูล

````python
actual_records = hill_daily_rain_filtered.groupby('station_code').size().reset_index(name='actual_records')
display(actual_records.head())
````

### ขั้นตอนที่ 4: คำนวณบันทึกที่หายไปและสัดส่วนความสมบูรณ์ของข้อมูล

**เหตุผล:** หลังจากได้ข้อมูลจำนวนวันที่คาดหวังและจำนวนบันทึกข้อมูลจริงของแต่ละสถานีแล้ว ขั้นตอนนี้จะทำการรวมข้อมูลทั้งสองเข้าด้วยกันเพื่อคำนวณจำนวนข้อมูลที่หายไปและแสดงเป็นสัดส่วนร้อยละความสมบูรณ์ของข้อมูล

````python
completeness_analysis = pd.merge(date_ranges, actual_records, on='station_code', how='left')
completeness_analysis['missing_records'] = completeness_analysis['expected_days'] - completeness_analysis['actual_records']
completeness_analysis['completeness_percentage'] = (completeness_analysis['actual_records'] / completeness_analysis['expected_days']) * 100

print("ผลการวิเคราะห์ความสมบูรณ์ของข้อมูลปริมาณน้ำฝนรายวัน (5 แถวแรก):\n")
display(completeness_analysis.head())
````

### ขั้นตอนที่ 5: แสดงผลสรุปความสมบูรณ์ของข้อมูล

**เหตุผล:** เพื่อให้เห็นภาพรวมของข้อมูลที่หายไปได้อย่างชัดเจน การแสดงสถานีที่มีสัดส่วนความสมบูรณ์ของข้อมูลต่ำที่สุดและสูงที่สุดจะช่วยให้สามารถระบุสถานีที่ต้องการการตรวจสอบเพิ่มเติมได้ง่ายขึ้น

````python
completeness_analysis_sorted = completeness_analysis.sort_values(by='completeness_percentage', ascending=True)
print("ผลการวิเคราะห์ความสมบูรณ์ของข้อมูลปริมาณน้ำฝนรายวัน (เรียงตามสัดส่วนความสมบูรณ์จากน้อยไปมาก, 5 แถวแรก):\n")
display(completeness_analysis_sorted.head())

print("\nผลการวิเคราะห์ความสมบูรณ์ของข้อมูลปริมาณน้ำฝนรายวัน (เรียงตามสัดส่วนความสมบูรณ์จากน้อยไปมาก, 5 แถวสุดท้าย):\n")
display(completeness_analysis_sorted.tail())
````

````python
# Identify stations with completeness percentage less than 80% (i.e., missing records >= 20%)
stations_to_drop = completeness_analysis[completeness_analysis['completeness_percentage'] < 80]

# Display the names of stations to be dropped
print("รายชื่อสถานีที่มีข้อมูลสูญหาย >= 20% ที่จะถูกคัดออก:")
display(stations_to_drop[['station_code', 'completeness_percentage']])

# Get the number of stations to be dropped
num_stations_to_drop = len(stations_to_drop)
print(f"\nจำนวนสถานีที่จะถูกคัดออก: {num_stations_to_drop} สถานี")

# Get the station codes of those to be dropped
dropped_station_codes = stations_to_drop['station_code'].tolist()

# Filter hill_daily_rain_filtered to keep only stations that are NOT in the dropped list
hill_daily_rain_cleaned = hill_daily_rain_filtered[~hill_daily_rain_filtered['station_code'].isin(dropped_station_codes)]

# Update completeness_analysis as well to reflect the removed stations
completeness_analysis_cleaned = completeness_analysis[~completeness_analysis['station_code'].isin(dropped_station_codes)]

# Display the number of remaining stations
num_remaining_stations = hill_daily_rain_cleaned['station_code'].nunique()
print(f"\nจำนวนสถานีที่เหลือหลังจากคัดออก: {num_remaining_stations} สถานี")

print("\nข้อมูล hill_daily_rain_cleaned (5 แถวแรก): ")
display(hill_daily_rain_cleaned.head())
````

## สร้างชุดข้อมูลวันที่มีอยู่จริงสำหรับแต่ละสถานี

### Subtask:
สำหรับแต่ละสถานี ให้สร้าง DataFrame ที่มีวันที่ทั้งหมดตั้งแต่ 'measure_datetime' ต่ำสุดถึงสูงสุดของสถานีนั้น เพื่อใช้เป็นฐานในการระบุว่าวันไหนที่ขาดหายไป


````python
date_ranges_per_station = hill_daily_rain_cleaned.groupby('station_code')['measure_datetime'].agg(['min', 'max'])

full_date_records = []
for station_code, row in date_ranges_per_station.iterrows():
    min_date = row['min']
    max_date = row['max']
    if pd.isna(min_date) or pd.isna(max_date):
        # Skip if min or max date is NaN (shouldn't happen with proper filtering but good practice)
        continue

    # Generate a full date range for the current station
    dates = pd.date_range(start=min_date, end=max_date, freq='D')

    # Append records for this station to the list
    for date in dates:
        full_date_records.append({'station_code': station_code, 'measure_datetime': date})

# Convert the list of dictionaries to a DataFrame
full_date_range = pd.DataFrame(full_date_records)

print("ข้อมูล 'full_date_range' ที่สร้างขึ้น (5 แถวแรก):")
display(full_date_range.head())
print("จำนวนแถวทั้งหมดใน full_date_range:", len(full_date_range))
print("ประเภทข้อมูลของ full_date_range:")
full_date_range.info()
````

### อภิปราย เหตุใดใช้ Median แทน mean

ในการเติมข้อมูลปริมาณน้ำฝนที่หายไป ผมเลือกใช้ ค่ามัธยฐาน (Median) แทน ค่าเฉลี่ย (Mean) ด้วยเหตุผลหลักๆ ดังนี้:

1.ความทนทานต่อค่าผิดปกติ (Robustness to Outliers): ข้อมูลปริมาณน้ำฝนมักจะมีค่าผิดปกติ (Outliers) เช่น วันที่ฝนตกหนักมากเป็นพิเศษ หรือวันที่ไม่มีฝนเลย (ค่าศูนย์) ซึ่งค่าเหล่านี้สามารถดึงค่าเฉลี่ยให้ผิดเพี้ยนไปได้มาก ในขณะที่ค่ามัธยฐานจะมีความทนทานต่อค่าผิดปกติเหล่านี้มากกว่า เนื่องจากเป็นค่ากลางที่แบ่งข้อมูลออกเป็นสองส่วนเท่าๆ กัน จึงได้รับผลกระทบจากค่าสุดโต่งน้อยกว่า

2.การกระจายของข้อมูลที่ไม่เป็นแบบปกติ (Skewed Distribution): ข้อมูลปริมาณน้ำฝนมักจะมีการกระจายตัวที่ไม่สมมาตร (Skewed) คือมีวันที่มีฝนตกน้อยหรือไม่มีเลยจำนวนมาก และมีวันที่ฝนตกหนักจำนวนน้อยกว่า สำหรับข้อมูลที่มีการกระจายตัวแบบนี้ ค่ามัธยฐานจะสะท้อนค่ากลางของข้อมูลได้ดีกว่าและแม่นยำกว่าค่าเฉลี่ย ซึ่งอาจถูกดึงไปทางค่าที่สูงกว่าในกรณีที่มีฝนตกหนักมากๆ

3.ความหมายทางกายภาพ (Physical Meaning): การใช้ค่ามัธยฐานช่วยให้มั่นใจได้ว่าค่าที่เติมเข้าไปนั้นเป็นค่าที่ 'ทั่วไป' หรือ 'ปกติ' ที่สุดของปริมาณน้ำฝนสำหรับวันนั้นๆ และสถานีนั้นๆ โดยไม่ถูกบิดเบือนจากเหตุการณ์ฝนตกหนักที่เกิดขึ้นไม่บ่อยนัก ทำให้การประมาณค่ามีความสมเหตุสมผลมากขึ้น  

การใช้ Median จึงเป็นวิธีการที่ "ปลอดภัยกว่า" (Conservative) และ "ทนทาน" (Robust) ต่อข้อมูลที่ผิดปกติ เหมาะสำหรับการสร้างข้อมูลตัวแทน (Representative Data) เพื่อนำไปรันโมเดลที่ไม่ต้องการให้ค่า Extreme มาบิดเบือนผลลัพธ์ในภาพรวม

## รวมข้อมูลฝนรายวันปี 2025 (เดือน 7-12)

### Subtask:
อ่านไฟล์ CSV ทั้งหมดจากโฟลเดอร์ `/content/drive/MyDrive/midterm.2/07_2025` ถึง `/content/drive/MyDrive/midterm.2/12_2025` และรวมเป็น DataFrame เดียวชื่อ `combined_daily_rain_2025`


**Reasoning**:
I need to consolidate all daily rainfall data for 2025 (months 7-12) into a single DataFrame. The previous cells read files from individual month folders one by one, overwriting `combined_daily_rain_2025` each time. To correctly combine all data, I will create a list of all relevant folder paths, iterate through them, read all CSV files within each folder, and then concatenate all resulting DataFrames into one, named `combined_daily_rain_2025`. I will also print its head and info to verify the consolidation.



````python
import os
import pandas as pd

# Define the base path for the monthly folders
base_folder_path = '/content/drive/MyDrive/midterm.2/'

# Create a list of folder paths for months 07 to 12 of 2025
folder_paths = [os.path.join(base_folder_path, f'{month:02d}_2025') for month in range(7, 13)]

data_list = []

# Loop through each folder path
for folder_path in folder_paths:
    # List all files in the current folder
    all_files = os.listdir(folder_path)

    # Filter for CSV files
    csv_files = [f for f in all_files if f.endswith('.csv')]

    # Loop through each CSV file and read it into a DataFrame
    for filename in csv_files:
        file_path = os.path.join(folder_path, filename)
        try:
            df = pd.read_csv(file_path)
            data_list.append(df)
        except Exception as e:
            print(f"Error reading {filename}: {e}")

# Concatenate all DataFrames in the list into a single DataFrame
if data_list:
    combined_daily_rain_2025 = pd.concat(data_list, ignore_index=True)
    print("ข้อมูลที่รวมกันแล้ว (5 แถวแรก):\n")
    display(combined_daily_rain_2025.head())
    print("\nข้อมูลของ DataFrame ที่รวมกันแล้ว:")
    combined_daily_rain_2025.info()
else:
    print("ไม่พบไฟล์ CSV ที่จะนำมารวม หรือเกิดข้อผิดพลาดในการอ่านไฟล์")
````

**Reasoning**:
I need to process the combined daily rainfall data for 2025. This involves converting the 'measure_datetime' column to a datetime format, extracting the month and year, calculating the monthly rainfall totals, and then renaming the columns to match the target format of 'Node', 'rain', 'month', 'year'. Finally, I will display the first few rows and the data types of the resulting DataFrame for verification.



````python
import pandas as pd

# Convert 'measure_datetime' to datetime objects
# Using errors='coerce' to handle any unparseable dates by setting them to NaT
combined_daily_rain_2025['measure_datetime'] = pd.to_datetime(combined_daily_rain_2025['measure_datetime'], errors='coerce')

# Drop rows where 'measure_datetime' could not be converted (if any)
combined_daily_rain_2025.dropna(subset=['measure_datetime'], inplace=True)

# Create 'month' and 'year' columns
combined_daily_rain_2025['month'] = combined_daily_rain_2025['measure_datetime'].dt.month
combined_daily_rain_2025['year'] = combined_daily_rain_2025['measure_datetime'].dt.year

# Calculate monthly rainfall totals
# Group by 'tele_station_oldcode', 'year', 'month' and sum 'rainfall_value'
monthly_rainfall_2025 = combined_daily_rain_2025.groupby(['tele_station_oldcode', 'year', 'month'])['rainfall_value'].sum().reset_index()

# Rename columns to match the desired format ('Node', 'rain', 'month', 'year')
monthly_rainfall_2025.rename(columns={'tele_station_oldcode': 'Node', 'rainfall_value': 'rain'}, inplace=True)

print("ข้อมูลปริมาณน้ำฝนรายเดือนปี 2025 (5 แถวแรก):")
display(monthly_rainfall_2025.head())
print("\nประเภทข้อมูลของ DataFrame ปริมาณน้ำฝนรายเดือน:")
monthly_rainfall_2025.info()
````

### บันทึกผลลัพธ์ปริมาณน้ำฝนรายเดือนปี 2025 ลงในไฟล์ CSV

````python
output_path = '/content/drive/MyDrive/midterm.2/monthly_rainfall_2025.csv'
monthly_rainfall_2025.to_csv(output_path, index=False)
print(f"บันทึกข้อมูลปริมาณน้ำฝนรายเดือนลงใน: {output_path} เรียบร้อยแล้ว")
````
